###  What Is an Array?

An array is a data structure that stores elements of the same type in a contiguous block of memory. In Python, NumPy provides the ndarray (n-dimensional array) object, which is more efficient and powerful than regular Python lists.
  





```python
#Creating Arrays
import numpy as np

# 1D array
a = np.array([1, 2, 3])

# 2D array
b = np.array([[1, 2], [3, 4]])

# Array of zeros
zeros = np.zeros((2, 3))

# Array of ones
ones = np.ones((3, 2))

# Range of numbers
r = np.arange(0, 10, 2)

# Evenly spaced numbers
lin = np.linspace(0, 1, 5)
```

###attributes of numpy

```python
a.shape       # Dimensions (e.g., (3,))
a.ndim        # Number of dimensions
a.size        # Total number of elements
a.dtype       # Data type of elements
```

##Array Operations

```python
#Element-wise Operations
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

a + b         # [5, 7, 9]
a * b         # [4, 10, 18]
a ** 2        # [1, 4, 9]
```

```python
#📊 Aggregation Functions
np.sum(a)     # Sum of elements
np.mean(a)    # Average
np.std(a)     # Standard deviation
np.max(a)     # Maximum
np.min(a)     # Minimum
```

###Indexing and Slicing

```python
#📌 Accessing Elements
a = np.array([[1, 2, 3], [4, 5, 6]])

a[0, 1]       # Element at row 0, column 1 → 2
a[:, 1]       # All rows, column 1 → [2, 5]
a[1, :]       # Row 1 → [4, 5, 6]
```

```python
#✂️ Slicing
a[0:2, 1:3]   # Subarray from rows 0–1 and columns 1–2
```

###🔄 Reshaping and Broadcasting

```python
#🔧 Reshape
a = np.array([1, 2, 3, 4, 5, 6])
a.reshape((2, 3))  # Converts to 2x3 matrix
```

```python
#📡 Broadcasting
#Allows operations between arrays of different shapes.

a = np.array([[1], [2], [3]])
b = np.array([10, 20, 30,40])
a + b  # Broadcasts b across rows of a
```

###🧹 Masking and Filtering

```python
#🎭 Boolean Indexing
a = np.array([1, 2, 3, 4, 5])
mask = a > 3
a[mask]       # [4, 5]
```

```python
#🐍 Python List
lst = [1, 2, 3]
result = [x * 2 for x in lst]  # Manual loop
```

```python
#🧠 NumPy Array
import numpy as np
arr = np.array([1, 2, 3])
result = arr * 2  # Vectorized operation
```

```python
#Performance Benchmark
import time
import numpy as np

size = 10**6
lst = list(range(size))
arr = np.array(lst)

# List timing
start = time.time()
lst_result = [x * 2 for x in lst]
print("List time:", time.time() - start)

# NumPy timing
start = time.time()
arr_result = arr * 2
print("NumPy time:", time.time() - start)
```

**The numpy.abs()** function calculates the absolute value of each element in a NumPy array.
*  It's a fundamental mathematical function that operates element-wise.
* You can also use the shorthand np.absolute().

**Key Features**
Element-wise operation: It applies the absolute value calculation to every element independently, returning a new array with the same shape as the input.

**Data Types:**
It works with integers, floating-point numbers, and complex numbers. For complex numbers, it returns the magnitude (Euclidean distance from the origin in the complex plane).

Broadcasting: It supports NumPy's broadcasting rules, allowing you to perform operations between arrays of different shapes under certain conditions.





```python
import numpy as np

arr = np.array([-5, 10, -15, 20])
abs_arr = np.abs(arr)
print(abs_arr)
```

np.all() is a powerful NumPy function used to check whether all elements in an array satisfy a condition.

Returns True if every element in the array evaluates to True; otherwise, returns False.

```python
import numpy as np

# A boolean array
a = np.array([False, False, False, False])
result = np.all(a)
print(result)

```

```python
# A numerical array (all non-zero, so they are treated as True)
b = np.array([1, 2, 3, 4])
result = np.all(b)
print(result)
```

```python
#With Conditions

x = np.array([2, 4, 6])
np.all(x > 1)     # True
np.all(x < 5)     # False
```

```python
#Across Rows or Columns
m = np.array([[1, 2, 3],
              [4, 5, 6]])

np.all(m > 0, axis=0)  # Column-wise → [True, True, True]
np.all(m > 2, axis=1)  # Row-wise → [False, True]

```

###np.append()

np.append() is a NumPy function used to add elements to an existing array

np.append(arr, values, axis=None)




```python
#Append to 1D Array

import numpy as np

arr2d = np.array([[1, 2],
                  ])
new_row = np.array([[5, 6]])

result_row = np.append(arr2d, new_row, axis=1)

print(result_row)
```

```python
#Append to 2D Array (Flattened)
a = np.array([[1, 2], [3, 4]])
b = np.append(a, [5, 6])
# Output: array([1, 2, 3, 4, 5, 6])

```

```python
#Append Along Axis
a = np.array([[1, 2], [3, 4]])
b = np.append(a, [[5, 6]], axis=0)
# Output:
# [[1 2]
#  [3 4]
#  [5 6]]
```

np.append() does not modify the original array — it returns a new array.

If axis is specified, the shapes must be compatible.

Internally, np.append() uses np.concatenate(), so it's not the most efficient for repeated appends.

###np.apply_along_axis()
 is a powerful NumPy function that lets you apply a custom function along a specific axis of a multidimensional array.

 It's especially useful when you want to perform row-wise or column-wise operations without writing explicit loops.


 np.apply_along_axis(func1d, axis, arr, *args, **kwargs)

func1d: Function to apply to 1D slices.

axis: Axis along which to apply the function.

axis=0 → apply to columns

axis=1 → apply to rows

arr: Input array.

```python
import numpy as np

# A 2x3 array
arr = np.array([[10, 2, 8],
                [5, 12, 4],
                [12,15,5]])

# Apply np.median along axis 1 (rows)
result = np.apply_along_axis(np.median, 0, arr)

print(result)
```

```python
#np.argmax() is a super useful NumPy function for identifying the index of the maximum value in an array.

import numpy as np

arr = np.array([3, 1, 8, 2, 5])
max_index = np.argmax(arr)

print(max_index)
```

```python
import numpy as np

arr = np.array([10, 3, 7, 1, 9, 2])
# Find the indices that would partition the array around the 2nd smallest element (index 2)
# The indices of the two smallest elements (1 and 2) will be at the beginning.
indices = np.argpartition(arr, -2)
print(indices[-2:])
```
