# what is data pipeline and automation
A data pipeline is a set of processes for moving and transforming data from one system to another. Think of it as a series of steps that data flows through, where each step performs a specific operation like cleaning, filtering, or aggregating the data.


Automation in this context refers to using tools and scripts to automatically execute these data pipeline steps without manual intervention. This saves time, reduces errors, and allows for processing large volumes of data efficiently.

# why do we use pip install
* You use pip install to install packages from the Python Package Index (PyPI).
* PyPI is a repository of software for the Python programming language.
* Packages contain modules and code that extend the functionality of Python, allowing you to do things like work with data (pandas), create visualizations (seaborn), or schedule tasks (schedule).
* So, pip install is the standard way to add these external libraries to your Python environment so you can use them in your code.


# why do we use pipi install instead of import
That's a great question! The difference between pip install and import is about when and how you access code:

pip install: This is a command-line tool used before you run your Python code. It downloads and installs external libraries (like pandas or schedule) onto your system or into your Python environment. Think of it like adding new tools to your toolbox. You only need to install a library once.

import: This is a Python keyword used within your Python code. It brings the code from an already-installed library into your current script or notebook so you can use its functions and classes. Think of it like taking a tool out of your toolbox to use for a specific task. You need to import a library every time you want to use it in a new Python script or session.

So, you pip install a library to make it available on your system, and then you import it in your script to use it.

```python
import schedule, time
import pandas as pd

def pipeline():
  data = pd.read_excel("/content/irisexcel1.xlsx")
  data = data.drop_duplicates()
  data = data.fillna(method="ffill")
  print("Rows:",len(data))
  print("Mean Values:\n",data.mean(numeric_only= True))

schedule.every(0.10).minutes.do(pipeline)

while True:
  schedule.run_pending()
  time.sleep(1)
```

This code defines and schedules a data processing pipeline to run automatically using the schedule library.

Here's a breakdown:

**import schedule, time:** Imports the necessary libraries: schedule for scheduling tasks and time for pausing the script's execution.

**import pandas as pd:** Imports the pandas library for data manipulation.

**def pipeline()**:: Defines a function named pipeline that contains the data processing steps.

**data = pd.read_excel("/content/irisexcel1.xlsx")**: Reads data from an Excel file into a pandas DataFrame.

**data = data.drop_duplicates():** Removes duplicate rows from the DataFrame.

**data = data.fillna(method="ffill"):** Fills missing values using the forward-fill method. Note: The output shows a FutureWarning for using method with fillna. It's recommended to use data.ffill() instead in newer pandas versions.

**print("Rows:",len(data)):** Prints the number of rows in the cleaned DataFrame.

**print("Mean Values:\n",data.mean(numeric_only= True)):** Calculates and prints the mean of the numeric columns in the DataFrame.

**schedule.every(0.10).minutes.do(pipeline):** Schedules the pipeline function to run every 0.10 minutes (6 seconds).

**while True:**: Starts an infinite loop to keep the script running.
schedule.run_pending(): Checks if any scheduled tasks are due to run and executes them.

**time.sleep(1):** Pauses the loop for 1 second.

This setup allows the pipeline function to automatically execute every 6 seconds, performing the data cleaning and analysis steps defined within it. The while True loop with schedule.run_pending() and time.sleep(1) is a common pattern for keeping a script alive and checking for scheduled tasks.

```python
import pandas as pd
import random
```

```python
names = ["Alice","Lekshmi","Sandra","Aswathy","Bob","James","Martin"]
department = ["Computer Science","Electronics","Mechanical","Civil"]

data_list = []
for i in range(1,21):
  record = {
      "student_id": i,
      "Names": random.choice(names),
      "department": random.choice(department+["Electrnics"]), # typo possiblity
      "age": random.choice([20,21,22,23,None]), # missing values possible
      "marks": random.randint(20,100),
      "attendance": random.choice([random.randint(70,100),None]) # some missing
  }
  data_list.append(record)

df = pd.DataFrame(data_list)

# save
df.to_csv("students_random.csv",index = True)
print("Random Dirty dataset saves as students_random.csv")
print(df.head(10))
```

This code generates a random dataset of student information, saves it as a CSV file, and then reads it back into a pandas DataFrame. After reading the data, it performs several data cleaning operations:

**Generate Random Data:** Creates a list of dictionaries, where each dictionary represents a student record with randomly selected values for Name, Department, Age, Marks, and Attendance. There is a potential typo in the "department" list ('Electrnics'). Missing values are intentionally included in 'age' and 'attendance'.
Create and Save DataFrame: Converts the list of records into a pandas DataFrame and saves it to a CSV file named "students_random.csv" with an index column.

Load Data: Reads the "students_random.csv" file back into a DataFrame.

Print Original Data: Displays the first 10 rows of the original dirty data.

Remove Duplicates: Removes any duplicate rows from the DataFrame.

Handle Missing Values:
Fills missing values in the 'age' column with the mean of the existing 'age' values.

Fills missing values in the 'attendance' column with the median of the existing 'attendance' values.

Fix Typo: Replaces the misspelled 'Electrnics' in the 'department' column with 'Electronics'.

Print Cleaned Data: Displays the first 10 rows of the cleaned DataFrame.

This cell demonstrates a basic data cleaning pipeline, including handling duplicates, missing values, and typos.

```python
df = pd.read_csv("students_random.csv")

print("Orginal Dirty Data")
print(df.head(10))

#1.Drop duplicate
df = df.drop_duplicates()

#2.handle missing values
df['age'] = df['age'].fillna(df['age'].mean())
df['attendance'] = df['attendance'].fillna(df['attendance'].median())

#3.fix typo in department
df['department'] = df['department'].replace({'Electrnics':'Electronics'})

print("\nCleaned Data")
print(df.head(10))
```

```python
# Automation rule -> check low marks
if "marks" in df.columns:
  low_marks = df[df["marks"]<40]
  if not low_marks.empty:
    print("Warning! students with marks below 40:")
    print(low_marks[['student_id','Names','marks']])
  else:
    print("No failing student found")
```

This code implements an automation rule to check for students with low marks in the df DataFrame.

Here's a breakdown:

if "marks" in df.columns:: This checks if a column named "marks" exists in the DataFrame. This is a good practice to prevent errors if the column is missing.

low_marks = df[df["marks"]<40]: This line filters the DataFrame to select only the rows where the value in the "marks" column is less than 40. These rows are stored in a new DataFrame called low_marks.

if not low_marks.empty:: This checks if the low_marks DataFrame is not empty, meaning there are students with marks below 40.

print("Warning! students with marks below 40:"): If there are students with low marks, this line prints a warning message.

print(low_marks[['student_id','Names','marks']]): This line prints the 'student_id', 'Names', and 'marks' columns for the students with low marks.

else:: If the low_marks DataFrame is empty (no students with marks below 40), the code inside this block is executed.

print("No failing student found"): This line prints a message indicating that no students with marks below 40 were found.

This cell demonstrates a simple automation rule that can be part of a larger data pipeline to flag or process data based on specific conditions.
