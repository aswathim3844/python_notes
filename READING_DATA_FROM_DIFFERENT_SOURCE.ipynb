## Reading and Handling Data from Different Sources

This notebook demonstrates how to read and handle data from various sources using the pandas library in Python.

### 1. Reading Data from Excel Files (.xlsx)

-   **`pd.read_excel(file_path, header, sheet_name)`**: This function is used to read data from Excel files.
    -   `file_path`: The path to the Excel file.
    -   `header`: Specifies the row (0-indexed) to use as the column headers.
    -   `sheet_name`: Specifies the sheet(s) to read. It can be a string (single sheet name), an integer (sheet index), a list of strings/integers, or `None` (reads all sheets into a dictionary of DataFrames).

-   **Reading all sheets**: When `sheet_name=None`, `pd.read_excel` returns a dictionary where keys are sheet names and values are DataFrames. You can access individual DataFrames using the sheet name as the key (e.g., `employees['employes']`).

-   **Concatenating DataFrames**: `pd.concat([df1, df2])` is used to combine DataFrames vertically (stacking them on top of each other). By default, it keeps the original index. Use `ignore_index=True` to reset the index.

### 2. Reading Data from CSV Files (.csv)

-   **`pd.read_csv(file_path)`**: This is the primary function for reading data from CSV files. It automatically infers delimiters in many cases.

### 3. Reading Data from XML Files (.xml)

-   **`pd.read_xml(file_path)`**: This function is used to parse XML files and load the data into a DataFrame. It requires the `lxml` library to be installed (`pip install lxml`). Ensure your XML file is well-formed.

### 4. Reading Data from JSON Files (.json)

-   **`pd.read_json(file_path)`**: This function reads data from JSON files. It can handle various JSON structures.

-   **Handling Nested Data (Exploding Lists)**: If a column contains lists (like the 'genres' column in the movies data), you can use the `.explode()` method to create a new row for each element in the list, while duplicating the values from other columns.

### 5. Reading Data from HTML Tables (.html or URL)

-   **`pd.read_html(url_or_file_path)`**: This function reads HTML tables from a URL or a local file path and returns a list of DataFrames, one for each table found.

-   **Accessing Specific Tables**: You can access individual tables from the list using their index (e.g., `tables[0]`).

-   **Combining Tables**: Similar to Excel sheets, you can use `pd.concat()` to combine multiple tables read from HTML.

### 6. Reading Data from APIs (JSON Response)

-   **Using the `requests` library**:
    -   `import requests`: Imports the library.
    -   `response = requests.get(url)`: Sends a GET request to the specified URL.
    -   `data = response.json()`: Parses the JSON response into a Python dictionary or list.

-   **Converting JSON to DataFrame**: `pd.DataFrame(data)` can directly convert a list of dictionaries (which is often the structure of a JSON API response) into a pandas DataFrame.

### 7. Data Handling and Manipulation

-   **Displaying Data**:
    -   `df.head()`: Shows the first few rows.
    -   `df.info()`: Provides a summary of the DataFrame, including column data types and non-null values.
    -   `df.shape`: Returns a tuple representing the dimensions of the DataFrame (rows, columns).
    -   `display(df)`: Provides a richer display of DataFrames in environments like Colab.

-   **Selecting Columns**: You can select one or more columns using square brackets (e.g., `df['column_name']` or `df[['col1', 'col2']]`).

-   **Filtering Data**: Use boolean indexing to filter rows based on conditions (e.g., `df[df['column'] > value]`).

-   **Grouping and Aggregating**:
    -   `df.groupby('column')`: Groups the DataFrame by unique values in a column.
    -   `.mean()`, `.count()`, `.sum()`, etc.: Apply aggregation functions to the grouped data.

-   **Sorting Data**: `df.sort_values('column', ascending=False)` sorts the DataFrame by a specific column.

-   **Renaming Columns**: `df.rename(columns={'old_name': 'new_name'}, inplace=True)` renames columns. Using `inplace=True` modifies the DataFrame directly.

-   **Merging DataFrames**: `df1.merge(df2, left_on='col1', right_on='col2', how='merge_type')` combines DataFrames based on common columns.
    -   `left_on`, `right_on`: Specify the columns to merge on in the left and right DataFrames.
    -   `how`: Specifies the type of merge ('inner', 'left', 'right', 'outer').

### Additional Tips

-   **Error Handling**: Be prepared for potential errors when reading files (e.g., `FileNotFoundError`, parsing errors). Using `try-except` blocks can make your code more robust.
-   **Data Cleaning**: Real-world data often requires cleaning (handling missing values, inconsistent formats, etc.) before analysis.
-   **Large Files**: For very large files, consider reading data in chunks to avoid memory issues.
-   **Documentation**: Refer to the pandas documentation for detailed information on all functions and methods.
